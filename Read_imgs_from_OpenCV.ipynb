## Reading images
We'll start by simply reading an image from a file. Next we go about "filming" our screen. We can use this for analyzing the images we see on the screen (we'll get into this later). Lastly we'll start reading images from our webcam and from a video file.

**1. Loading a file**

First we import OpenCV on line 1 (cv2 is OpenCV's module name in the Python implementation).

```python
import cv2
```

```python
img = cv2.imread(filename="C:/Users/Usuario/GIT/Procesamiento_de_imagenes_UC/landscape.png")
cv2.imshow('image', img) # muestra la imagen
cv2.waitKey(0) # al precionar cualquier tecla se cierra con la siguiente linea
cv2.destroyAllWindows() 
```

```python
img = cv2.imread(filename="C:/Users/Usuario/GIT/Procesamiento_de_imagenes_UC/landscape.png")
img = cv2.cvtColor(src=img, code=cv2.COLOR_BGR2RGB) # por defecto cv2 deja las imagenes en este formato BGR que es como una escala de grices se pasa a RGB para dejarla a color
cv2.imshow('image', img)
cv2.waitKey(0)
cv2.destroyAllWindows() 
```

**2. Filming your screen**

Want to create an AI for a game or analyze video data? Then you might want to read your screen in order to analyze the images. It's quite similar to loading a file:

```python
import numpy as np
from PIL import ImageGrab
```

```python
while (True):
    screen = np.array(ImageGrab.grab()) # se convierte a un array de numpy para que OpenCV pueda trabajar con el
    screen = cv2.cvtColor(src=screen, code=cv2.COLOR_BGR2RGB)
    cv2.imshow('my_screen', screen)

    # press escape to exit
    if (cv2.waitKey(30) == 27):
       break
cv2.destroyAllWindows()
```

**3. Reading th webcam**

```python
cap = cv2.VideoCapture(0) # cv2 captura la webcam
while True:
    ret, frame = cap.read() # leemos dos variables "ret" que nos dice si podemos leer frames y "frame"
    print("ret", ret, "frame", frame)
    frame = cv2.cvtColor(src=frame, code=cv2.COLOR_BGR2RGB)
    cv2.imshow('webcam', frame)

    # press escape to exit
    if (cv2.waitKey(30) == 27):
       break
cap.release()
cv2.destroyAllWindows()
```
